```python
# Preparação do ambiente
import pandas as pd

# ---------------------------------------------------------------
# Origem dos dados: repositório público no GitHub
# ---------------------------------------------------------------
# O CSV é lido por URL para que este notebook rode sozinho, em qualquer pasta,
# sem precisar do restante do projeto. O pandas aceita URL diretamente em
# read_csv.
#
# Esta é a primeira célula de código do notebook, então é aqui que as
# constantes de origem dos dados ficam definidas. A subseção 6.0 reaproveita
# estas mesmas constantes, em vez de repetir a URL em dois lugares.
#
# Para trabalhar offline ou com uma cópia local, troque BASE_DADOS pelo
# caminho relativo da pasta do projeto:
#     BASE_DADOS = "../../"
#
# Nota de reprodutibilidade: a URL aponta para o branch "main", então o
# notebook sempre lê a versão mais recente dos dados. Para congelar a versão
# usada nesta análise, troque "main" pelo hash do commit.
BASE_DADOS = "https://raw.githubusercontent.com/valandro/ufrgs-spec-ai/main/AI001-analise-visualizacao-dados/"

URL_DATASET_PRINCIPAL = BASE_DADOS + "educacao_rs/datasets/dataset_principal.csv"

# Carregamento dos dados
df = pd.read_csv(URL_DATASET_PRINCIPAL)

# Exploração inicial dos dados
print("INFORMAÇÕES DA BASE DE DADOS")
print("=" * 70)

print(f"Número de registros: {df.shape[0]}")
print(f"Número de atributos: {df.shape[1]}")

print("\nRESUMO DOS REGISTROS")
print("=" * 70)

print(df.head())
```

**Base de Dados:** Não foi aplicada nenhuma amostragem, a análise considera a totalidade dos 496 municípios do Rio Grande do Sul presentes na base, além dos registros agregados de Brasil e Rio Grande do Sul (usados apenas como referência, não incluídos na análise por unidade municipal).

```python
# As duas primeiras linhas correspondem ao Brasil e ao Rio Grande do Sul.
# Elas serão excluídas da análise das unidades individuais.
df_municipios = df.iloc[2:].reset_index(drop=True)

# Verificar colunas totalmente vazias
colunas_vazias = df_municipios.columns[df_municipios.isna().all()].tolist()

# Resumo
print("\nATRIBUTOS COMPLETAMENTE VAZIOS (MUNICÍPIOS)")
print("=" * 70)

print(f"Total de registros para análise (excluindo registros do Brasil e RS): {df_municipios.shape[0]}")
print(f"Total de atributos: {df_municipios.shape[1]}")
print(f"Atributos totalmente vazios (municípios): {len(colunas_vazias)}")

# filtrar as colunas totalmente vazias do DataFrame
df_municipios = df_municipios.drop(columns=colunas_vazias)
print(f"Total de atributos com algum dado (municípios): {df_municipios.shape[1]}")
```

```python
#Filtrar os atributos com percentual de nulos maior que 90%
df_filtrado = df_municipios.loc[:, (df_municipios.isna().sum() / len(df_municipios)) * 100 <= 95]

resumo = []

print("CARACTERÍSTICAS DOS ATRIBUTOS COM PERCENTUAL DE NULOS <= 95%")
print("=" * 70)

for coluna in df_filtrado.columns:

    tipo = df_filtrado[coluna].dtype
    nulos = df_filtrado[coluna].isna().sum()
    percentual_nulos = (nulos / len(df_filtrado)) * 100

    resumo.append({
        "Atributo": coluna,
        "Tipo": str(tipo),
        "Nulos": nulos,
        "% nulos": round(percentual_nulos, 2)
    })

resumo_df = pd.DataFrame(resumo)

#ordenar o DataFrame pelo percentual de nulos em ordem decrescente
resumo_df = resumo_df.sort_values(by="% nulos", ascending=False)
print("\n")
print(resumo_df.to_string(index=False))

print(f"\nNúmero de registros filtrados: {df_filtrado.shape[0]}")
print(f"Número de atributos filtrados: {df_filtrado.shape[1]}")
```

```python
colunas_interesse = [
    "Territorialidades",
    "% dos ocupados com ensino médio completo 2010",
    "Rendimento médio dos ocupados 2010",
    "% dos ocupados sem rendimento 2010",
    "% dos ocupados com rendimento de até 1 salário mínimo 2010",
    "% dos ocupados com rendimento de até 2 salários mínimo 2010",
    "% dos ocupados com rendimento de até 3 salários mínimo 2010",
    "% dos ocupados com rendimento de até 5 salários mínimo 2010"
]

df_filtrado = df_filtrado[colunas_interesse]

# Transformar a coluna "% dos ocupados com ensino médio completo 2010" em numérica
df_filtrado["% dos ocupados com ensino médio completo 2010"] = pd.to_numeric(
    df_filtrado["% dos ocupados com ensino médio completo 2010"], errors="coerce"
)

colunas_com_nulos = [
    "% dos ocupados com rendimento de até 1 salário mínimo 2010",
    "% dos ocupados sem rendimento 2010",
    "Rendimento médio dos ocupados 2010",
    "% dos ocupados com ensino médio completo 2010",
]

print("\nREGISTROS COM NULOS NOS ATRIBUTOS DE INTERESSE")
print("=" * 70)

linhas_nulas = df_municipios[df_municipios[colunas_com_nulos].isnull().any(axis=1)][["Territorialidades"] + colunas_com_nulos]
print(linhas_nulas.to_string(index=False))

```

```python
# Filtrar os registros com nulos nos atributos de interesse
df_filtrado = df_filtrado.dropna(subset=colunas_com_nulos).reset_index(drop=True)

print("ESTATÍSTICAS DOS ATRIBUTOS DE INTERESSE")
print("=" * 70)
print(f"\nMunicípios para análise: {len(df_filtrado)}")

print(df_filtrado.describe().T.to_string())
```

```python
import unicodedata

# Checagem de duplicidades em df_filtrado

# 1. Linhas 100% idênticas (todas as colunas iguais)
linhas_duplicadas = df_filtrado[df_filtrado.duplicated(keep=False)]
if len(linhas_duplicadas) > 0:
    print(linhas_duplicadas[["Territorialidades"]].to_string())

# 2. Nome de município repetido (comparação exata)
contagem_nomes = df_filtrado["Territorialidades"].value_counts()
nomes_repetidos = contagem_nomes[contagem_nomes > 1]
if len(nomes_repetidos) > 0:
    print(nomes_repetidos)

# 3. Nome "quase" igual — normaliza (tira acento, espaço extra, maiúscula/minúscula)
#    e compara de novo. Pega erro de digitação que a comparação exata não pega.
def normalizar(texto):
    texto = str(texto).strip().lower()
    texto = unicodedata.normalize("NFKD", texto).encode("ascii", "ignore").decode("ascii")
    return " ".join(texto.split())

df_filtrado = df_filtrado.copy()
df_filtrado["_nome_normalizado"] = df_filtrado["Territorialidades"].apply(normalizar)
contagem_normalizado = df_filtrado["_nome_normalizado"].value_counts()
possiveis_variacoes = contagem_normalizado[contagem_normalizado > 1]

if len(possiveis_variacoes) > 0:
    for nome_norm in possiveis_variacoes.index:
        variantes = df_filtrado[df_filtrado["_nome_normalizado"] == nome_norm]["Territorialidades"].tolist()
        print(f"  {nome_norm!r} aparece como: {variantes}")

df_filtrado = df_filtrado.drop(columns=["_nome_normalizado"])

# 4. Resumo
print("ANÁLISE DE DUPLICIDADES")
print("=" * 60)
print(f"Linhas totalmente idênticas:              {len(linhas_duplicadas)}")
print(f"Nomes de município exatamente repetidos:   {len(nomes_repetidos)}")
print(f"Possíveis variações do mesmo nome:         {len(possiveis_variacoes)}")
if len(linhas_duplicadas) == 0 and len(nomes_repetidos) == 0 and len(possiveis_variacoes) == 0:
    print("\nNenhuma duplicidade encontrada nos 3 níveis checados.")
```

```python
import matplotlib.pyplot as plt

col_escolaridade = "% dos ocupados com ensino médio completo 2010"
col_renda = "Rendimento médio dos ocupados 2010"

atributos = [
    (col_escolaridade, "% de ocupados com ensino médio completo (2010)", "%"),
    (col_renda, "Rendimento médio dos ocupados (2010)", "R$"),
]
 
fig, axes = plt.subplots(1, 2, figsize=(11, 4.5))
 
for ax, (coluna, titulo, unidade) in zip(axes, atributos):
    serie = df_filtrado[coluna].dropna()
 
    contagens, _, _ = ax.hist(serie, bins=20, edgecolor="white", linewidth=0.6, zorder=3)
 
    media = serie.mean()
    ax.axvline(media, color="#0b0b0b", linestyle="--", linewidth=1, alpha=0.6, zorder=4)
 
    # headroom acima da barra mais alta, pra caber o rótulo da média sem sobrepor
    y_max = contagens.max()
    ax.set_ylim(0, y_max * 1.18)
 
    rotulo = f"R$ {media:,.2f}" if unidade == "R$" else f"{media:.1f}%"
    ax.text(
        media, y_max * 1.05,
        f" média = {rotulo}",
        fontsize=8.5, color="#3a3a38", va="bottom", ha="left",
    )
 
    ax.set_title(titulo, fontsize=11, pad=10)
    ax.set_xlabel(unidade if unidade == "%" else "R$")
    ax.set_ylabel("Nº de municípios")
 
    ax.spines[["top", "right"]].set_visible(False)
    ax.set_axisbelow(True)
    ax.grid(axis="y", linestyle="-", linewidth=0.5, zorder=0)
 
fig.suptitle(
    "Distribuição dos municípios do RS - escolaridade e renda dos ocupados",
    fontsize=13, y=1.03,
)
 
plt.tight_layout()
plt.show()
```

Os dois histogramas apresentam distribuição unimodal e assimetria à direita, indicando maior concentração dos municípios em valores baixos ou intermediários e uma quantidade menor de municípios com valores mais elevados.

Na escolaridade, a maior concentração está entre 20% e 40% de ocupados com ensino médio completo. A média (29,7%) e a mediana (29,1%) são próximas, indicando uma assimetria relativamente moderada.

Na renda, os valores estão mais concentrados entre R\$ 750 e R\$ 1.250. A média (R\$ 956,61) é superior à mediana (R\$ 936,76), evidenciando uma influência maior dos municípios com rendimentos elevados, que chegam a aproximadamente R\$ 2.250.

De forma geral, os histogramas indicam maior dispersão da renda entre os municípios do que da escolaridade, com a presença de alguns municípios que se afastam da concentração observada.

```python
# código do gráfico da pergunta 1
col_escolaridade = "% dos ocupados com ensino médio completo 2010"
col_renda = "Rendimento médio dos ocupados 2010"
 
for col in [col_escolaridade, col_renda]:
    municipios[col] = pd.to_numeric(municipios[col], errors="coerce")
 
municipios = municipios.dropna(subset=[col_escolaridade, col_renda])
print(f"Municípios com os dois indicadores preenchidos: {len(municipios)}")
 
# ---------------------------------------------------------------
# 2. Estatística de apoio (reta de tendência e correlação)
# ---------------------------------------------------------------
slope, intercept, r, p, se = stats.linregress(
    municipios[col_escolaridade], municipios[col_renda]
)
x_linha = pd.Series([municipios[col_escolaridade].min(), municipios[col_escolaridade].max()])
y_linha = slope * x_linha + intercept
 
# ---------------------------------------------------------------
# 3. Gráfico
# ---------------------------------------------------------------
COR_PONTOS = "#2a78d6"
COR_LINHA = "#52514e"
 
fig, ax = plt.subplots(figsize=(8, 6))
 
ax.scatter(
    municipios[col_escolaridade],
    municipios[col_renda],
    s=22,
    color=COR_PONTOS,
    alpha=0.6,
    edgecolors="none",
    label="Municípios do RS (n=%d)" % len(municipios),
)
ax.plot(x_linha, y_linha, color=COR_LINHA, linewidth=2, linestyle="--", label="Tendência linear")
 
ax.set_title(
    "Municípios com mais trabalhadores com ensino médio completo\npagam salários mais altos?",
    fontsize=13,
    pad=12,
)
ax.set_xlabel("Ocupados com ensino médio completo (%)")
ax.set_ylabel("Rendimento médio dos ocupados (R$, 2010)")
 
ax.spines[["top", "right"]].set_visible(False)
ax.grid(axis="y", linestyle="-", linewidth=0.5, color="#e5e5e0", zorder=0)
ax.set_axisbelow(True)
 
ax.annotate(
    f"r de Pearson = {r:.2f}",
    xy=(0.03, 0.94),
    xycoords="axes fraction",
    fontsize=10,
    color=COR_LINHA,
)
 
ax.legend(frameon=False, loc="lower right")
 
fig.text(
    0.01, -0.02,
    "Fonte: Atlas do Desenvolvimento Humano no Brasil (Pnud, Ipea, FJP), com dados do Censo 2010 (IBGE).",
    fontsize=8, color="#8a8a86",
)
 
# plt.tight_layout()
# plt.savefig("grafico1_escolaridade_renda.png", dpi=150, bbox_inches="tight")
plt.show()
 
 
```

#### 6.2 Pergunta 2 — Como muda a distribuição de faixas de renda conforme o nível de escolaridade do município?


<a id="sec-6-2-1"></a>
##### 6.2.1 Gráfico da pergunta 2 — quartis de escolaridade (`pd.qcut`)

Converte as faixas cumulativas em faixas mutuamente exclusivas, agrupa os municípios em **quartis**
pelo % de ocupados com ensino médio completo (mesma quantidade de municípios por grupo) e mostra a
distribuição média de faixas de rendimento em barras 100% empilhadas. É esta a versão usada como
resposta da pergunta 2.

```python
#Gráfico pergunta "Como muda a distribuição de faixas de renda conformeo nível de escolaridade do município?"

import numpy as np

col_escolaridade = "% dos ocupados com ensino médio completo 2010"
cols_faixas_cumulativas = [
    "% dos ocupados sem rendimento 2010",
    "% dos ocupados com rendimento de até 1 salário mínimo 2010",
    "% dos ocupados com rendimento de até 2 salários mínimo 2010",
    "% dos ocupados com rendimento de até 3 salários mínimo 2010",
    "% dos ocupados com rendimento de até 5 salários mínimo 2010",
]
for c in [col_escolaridade] + cols_faixas_cumulativas:
    municipios[c] = pd.to_numeric(municipios[c], errors="coerce")

municipios = municipios.dropna(subset=[col_escolaridade] + cols_faixas_cumulativas)

# ---------------------------------------------------------------
#    Conversão das faixas cumulativas em faixas mutuamente exclusivas
#    As colunas do Atlas são cumulativas (até_1 >= sem_rendimento,
#    até_2 >= até_1, etc.), e não faixas independentes. O assert logo
#    abaixo garante isso: se alguma subtração desse negativa, a premissa
#    estaria errada e a célula falharia em vez de gerar um gráfico falso.
# ---------------------------------------------------------------
nomes_faixas = ["Sem rendimento", "Mais de 0 até 1 SM", "Mais de 1 até 2 SM",
                 "Mais de 2 até 3 SM", "Mais de 3 até 5 SM", "Mais de 5 SM"]
 
# remove as colunas de faixa se a célula for rodada de novo, pra não duplicar
municipios = municipios.drop(columns=nomes_faixas, errors="ignore")

sem_rendimento, ate_1, ate_2, ate_3, ate_5 = [municipios[c] for c in cols_faixas_cumulativas]

faixas_exclusivas = pd.DataFrame({
    "Sem rendimento": sem_rendimento,
    "Mais de 0 até 1 SM": ate_1 - sem_rendimento,
    "Mais de 1 até 2 SM": ate_2 - ate_1,
    "Mais de 2 até 3 SM": ate_3 - ate_2,
    "Mais de 3 até 5 SM": ate_5 - ate_3,
    "Mais de 5 SM": 100 - ate_5,
})

assert (faixas_exclusivas.min() >= -0.01).all(), "faixa negativa encontrada — checar dados"
municipios = pd.concat([municipios, faixas_exclusivas], axis=1)

# ---------------------------------------------------------------
#    Agrupamento por faixa de escolaridade. Os 3 grupos ficam com o MESMO
#    intervalo de pontos percentuais, mas NÃO com a mesma quantidade
#    de municípios: como a maioria dos municípios do RS está
#    concentrada na faixa baixa/média de escolaridade, o grupo "alta"
#    acaba com poucos municípios — vale checar o n de cada grupo antes
#    de interpretar o gráfico.
# ---------------------------------------------------------------
municipios["faixa_escolaridade"], limites = pd.cut(
    municipios[col_escolaridade], 3,
    labels=["Baixa escolaridade", "Média escolaridade", "Alta escolaridade"],
    retbins=True,
)
print("\nLimites das faixas (%):", [round(b, 1) for b in limites])

nomes_faixas = list(faixas_exclusivas.columns)
medias_por_faixa = municipios.groupby("faixa_escolaridade", observed=True)[nomes_faixas].mean()
contagem_por_faixa = municipios.groupby("faixa_escolaridade", observed=True).size()


# ---------------------------------------------------------------
#  Gráfico — barras 100% empilhadas, rampa sequencial (1 tom, claro->escuro)
# ---------------------------------------------------------------
CORES_SEQUENCIAIS = ["#86b6ef", "#5598e7", "#2a78d6", "#256abf", "#184f95", "#0d366b"]

fig, ax = plt.subplots(figsize=(8, 6))

grupos = medias_por_faixa.index.tolist()
x = np.arange(len(grupos))
base = np.zeros(len(grupos))

for faixa, cor in zip(nomes_faixas, CORES_SEQUENCIAIS):
    valores = medias_por_faixa[faixa].values
    ax.bar(x, valores, bottom=base, color=cor, width=0.6, label=faixa)

    for i, v in enumerate(valores):
        if v >= 5:
            ax.text(
                x[i], base[i] + v / 2, f"{v:.0f}%",
                ha="center", va="center", fontsize=9,
                color="white" if faixa in ("Mais de 2 até 3 SM", "Mais de 3 até 5 SM", "Mais de 5 SM") else "#0b0b0b",
            )
    base += valores

ax.set_xticks(x)
ax.set_xticklabels([f"{g}\n(n={contagem_por_faixa[g]})" for g in grupos])
ax.set_ylim(0, 100)

ax.set_title(
    "Como muda a distribuição de faixas de renda\nconforme o nível de escolaridade do município? (faixas de largura fixa)",
    fontsize=13, pad=12,
)
ax.set_xlabel(f"Faixa de % de ocupados com ensino médio completo (largura fixa, cortes em {limites[1]:.0f}% e {limites[2]:.0f}%)")
ax.set_ylabel("Ocupados por faixa de rendimento (%)")

ax.spines[["top", "right"]].set_visible(False)
ax.set_axisbelow(True)

handles, labels = ax.get_legend_handles_labels()
handles, labels = handles[::-1], labels[::-1]
ax.legend(handles, labels, frameon=False, loc="center left", bbox_to_anchor=(1.02, 0.5), fontsize=9)


plt.tight_layout()
plt.show()
```

<a id="sec-6-2-2"></a>
##### 6.2.2 Interpretação

**Resposta curta: sim — a distribuição se desloca de forma consistente, e o deslocamento está na base, não no topo.**

- Do Q1 (12,4%–22,7% de escolaridade) ao Q4 (35,7%–64,5%), a fatia de ocupados **sem rendimento** cai de 18,2% para 4,8%, e a de **até 1 SM** cai de 28,5% para 13,3%. Somadas, as duas faixas mais baixas passam de **46,7% dos ocupados no Q1 para 18,1% no Q4**.
- Quase todo esse ganho é absorvido pela faixa de **1 a 2 SM**, que sobe de 36,3% para 50,5%. O que a escolaridade do município acompanha é a saída da base, e não a chegada ao topo.
- O topo se move pouco em termos absolutos: **mais de 5 SM** vai de 3,6% para 7,6%. Dobra em proporção, mas mesmo nos municípios mais escolarizados do estado são menos de 8 em cada 100 ocupados.
- O padrão é monótono nas seis faixas e nos quatro quartis — nenhum degrau inverte o sentido.
- **Por que os quartis (6.2.1) são a resposta, e não faixas de largura fixa:** com `pd.cut` o padrão se repete, mas os grupos ficam com 265, 210 e 21 municípios. Uma média calculada sobre 21 casos é frágil; os quartis mantêm cerca de 124 municípios em cada grupo.
- **Limites da leitura:** as faixas foram obtidas por subtração das colunas cumulativas, e a última (`mais de 5 SM`) por `100 − até 5 SM`, então ela acumula o arredondamento das anteriores. Além disso, a unidade é o município: esta é a média das distribuições municipais, e não a distribuição do estado, que daria mais peso aos municípios populosos.

- **O ranking das regiões muda**, porque os agrupamentos mudaram. A conclusão também — ver a interpretação em 6.3.4.

O cruzamento é por nome normalizado: `normalizar` (seção 5) resolve acento e pontuação, o que basta para `Westfália`/`Westfalia` e `Xangri-lá`/`Xangri-Lá`. Só `Santana` × `Sant'Ana do Livramento` precisa de equivalência explícita, registrada no código.

```python
# ---------------------------------------------------------------
# Pergunta 3 — preparacao: cruzamento com as Regioes Funcionais
# ---------------------------------------------------------------
# A fonte aqui e a tabela OFICIAL do Decreto 54.572/2019 (SEPLAG-RS), que
# associa cada um dos municipios do estado ao seu COREDE e a sua Regiao
# Funcional de Planejamento. Sao 28 COREDEs agrupados em 9 RFs.
#
# Uma versao anterior desta analise usava um arquivo que listava apenas os
# "principais municipios" de cada COREDE — 84 dos 496 — e que, alem de
# incompleto, agrupava os COREDEs em RFs de forma divergente da oficial.
# Com a tabela completa, a unidade de analise passa a ser o MUNICIPIO, e
# nao mais o polo regional: some o vies de selecao que limitava a leitura.
import numpy as np

col_renda = "Rendimento médio dos ocupados 2010"

regioes = pd.read_csv(URL_COREDE_RF)
print(f"Tabela oficial: {len(regioes)} municípios, "
      f"{regioes['corede'].nunique()} COREDEs, "
      f"{regioes['regiao_funcional_id'].nunique()} Regiões Funcionais")

base = df[~df["Territorialidades"].isin(["Brasil", "Rio Grande do Sul"])].copy()
base[col_renda] = pd.to_numeric(base[col_renda], errors="coerce")

# ---------------------------------------------------------------
# Cruzamento por nome normalizado.
# `normalizar` (seção 5) resolve acento e pontuação, o que basta para
# "Westfália/Westfalia" e "Xangri-lá/Xangri-Lá". Não resolve grafias
# historicamente diferentes: "Santana" x "Sant'Ana do Livramento" entra
# numa tabela de equivalências explícita, para ficar auditável.
# ---------------------------------------------------------------
EQUIVALENCIAS = {"Santana do Livramento (RS)": "Sant'Ana do Livramento (RS)"}
regioes["chave"] = regioes["Territorialidades"].replace(EQUIVALENCIAS).map(normalizar)
base["chave"] = base["Territorialidades"].map(normalizar)

sem_par = sorted(set(regioes["chave"]) - set(base["chave"]))
if sem_par:
    raise ValueError(f"Municípios da tabela oficial sem par no Atlas: {sem_par}. "
                     "Acrescente-os a EQUIVALENCIAS em vez de ignorá-los.")

p3 = base.merge(
    regioes[["chave", "municipio", "codigo_ibge", "corede", "regiao_funcional_id"]],
    on="chave", how="inner", validate="one_to_one",
).dropna(subset=[col_renda])

print(f"Municípios cruzados: {len(p3)} de {len(base)} do Atlas")
if len(p3) != 496:
    raise ValueError(f"Esperados 496 municípios no cruzamento, obtidos {len(p3)}.")

# Nomes descritivos das 9 RFs, montados a partir dos COREDEs que cada uma
# reúne — conferidos contra os dados logo abaixo.
NOMES_RF = {
    1: "RF1 · Metropolitana", 2: "RF2 · Vale do Rio Pardo e Taquari",
    3: "RF3 · Serra e Hortênsias", 4: "RF4 · Litoral", 5: "RF5 · Sul",
    6: "RF6 · Campanha e Fronteira Oeste", 7: "RF7 · Noroeste e Missões",
    8: "RF8 · Central", 9: "RF9 · Norte e Produção",
}
p3["rf_nome"] = p3["regiao_funcional_id"].map(NOMES_RF)
if p3["rf_nome"].isna().any():
    raise ValueError("Há Região Funcional sem nome em NOMES_RF.")

print("\nComposição de cada Região Funcional:")
for rf_id in sorted(p3["regiao_funcional_id"].unique()):
    fatia = p3[p3["regiao_funcional_id"] == rf_id]
    print(f"  {NOMES_RF[rf_id]:36s} {len(fatia):3d} municípios · "
          f"{', '.join(sorted(fatia['corede'].unique()))}")

# ---------------------------------------------------------------
# Estatisticas por Regiao Funcional
# ---------------------------------------------------------------
resumo_rf = (
    p3.groupby("rf_nome")
    .agg(n=(col_renda, "size"), coredes=("corede", "nunique"),
         mediana=(col_renda, "median"), media=(col_renda, "mean"),
         minimo=(col_renda, "min"), maximo=(col_renda, "max"))
    .sort_values("mediana", ascending=False)
)
print("\nRendimento médio dos ocupados por Região Funcional (496 municípios):")
print(resumo_rf.round(0).to_string())

posicao_rf1 = int(resumo_rf.index.get_loc(NOMES_RF[1])) + 1
lider = resumo_rf.index[0]
print(f"\nA RF1 (metropolitana) é a {posicao_rf1}ª de {len(resumo_rf)} em rendimento mediano.")
print(f"A primeira colocada é a {lider}.")

metro = p3.loc[p3["regiao_funcional_id"] == 1, col_renda]
demais = p3.loc[p3["regiao_funcional_id"] != 1, col_renda]
metro_sem_poa = p3.loc[(p3["regiao_funcional_id"] == 1) & (p3["municipio"] != "Porto Alegre"), col_renda]
serra = p3.loc[p3["regiao_funcional_id"] == 3, col_renda]

print("\nMetropolitana x demais:")
for rotulo, serie in [("RF1", metro), ("RF1 sem Porto Alegre", metro_sem_poa),
                      ("Demais regiões", demais), ("RF3 (Serra e Hortênsias)", serra)]:
    print(f"  {rotulo:26s} n={len(serie):3d}   mediana R$ {serie.median():,.0f}   "
          f"média R$ {serie.mean():,.0f}")

acima = int((demais > metro.median()).sum())
print(f"\nMunicípios não metropolitanos acima da mediana da RF1: {acima} de {len(demais)} "
      f"({100 * acima / len(demais):.0f}%)")

# ---------------------------------------------------------------
# Teste de permutacao: as diferencas resistem ao acaso?
# ---------------------------------------------------------------
# Embaralha os rotulos dos grupos e mede quao extrema e a diferenca
# observada. Nao supoe normalidade — as distribuicoes de renda municipal
# sao assimetricas.
N_PERMUTACOES = 20000
rng = np.random.default_rng(1)


def teste_permutacao(a, b):
    observada = np.median(a) - np.median(b)
    juntos = np.concatenate([np.asarray(a), np.asarray(b)])
    difs = np.array([
        np.median(p[:len(a)]) - np.median(p[len(a):])
        for p in (rng.permutation(juntos) for _ in range(N_PERMUTACOES))
    ])
    return observada, float((np.abs(difs) >= abs(observada)).mean())


print(f"\nTestes de permutação ({N_PERMUTACOES:,} reamostras):".replace(",", "."))
observada_rf1, p_valor = teste_permutacao(metro.values, demais.values)
print(f"  RF1 − demais regiões:      R$ {observada_rf1:+,.0f}   p = {p_valor:.4f}")
observada_serra, p_valor_serra = teste_permutacao(serra.values, metro.values)
print(f"  RF3 (Serra) − RF1:         R$ {observada_serra:+,.0f}   p = {p_valor_serra:.4f}")
```

##### 6.3.4 Interpretação

**Resposta: não. A proximidade da capital não explica o rendimento — e a região que mais paga no estado não é a metropolitana, é a Serra.**

**1. O ranking das 9 Regiões Funcionais**

| Região Funcional | Municípios | Mediana | Média |
|---|---|---|---|
| **RF3 · Serra e Hortênsias** | 48 | **R\$ 1.173** | R\$ 1.148 |
| **RF1 · Metropolitana** | 70 | **R\$ 997** | R\$ 1.011 |
| RF4 · Litoral | 21 | R\$ 995 | R\$ 1.017 |
| RF6 · Campanha e Fronteira Oeste | 20 | R\$ 979 | R\$ 972 |
| RF8 · Central | 49 | R\$ 920 | R\$ 941 |
| RF9 · Norte e Produção | 130 | R\$ 911 | R\$ 944 |
| RF2 · Vale do Rio Pardo e Taquari | 59 | R\$ 904 | R\$ 902 |
| RF7 · Noroeste e Missões | 77 | R\$ 799 | R\$ 875 |
| RF5 · Sul | 22 | R\$ 792 | R\$ 836 |

A metropolitana é a **2ª de 9**. Fica acima da maioria, mas atrás da Serra — e por uma margem grande.

**2. A Serra paga mais que a metropolitana, e a diferença é sólida**

A mediana da RF3 supera a da RF1 em **R\$ 176**. O teste de permutação com 20.000 reamostras dá **p = 0,0003**: uma diferença desse tamanho seria muito improvável se as duas regiões viessem da mesma distribuição.

Outras leituras na mesma direção: **69% dos municípios da Serra** (33 de 48) estão acima da mediana metropolitana, e entre os 20 municípios de maior rendimento do estado, **7 são da Serra contra 3 da metropolitana**.

**3. A vantagem da metropolitana sobre o resto é modesta e não conclusiva**

RF1 tem mediana R\$ 997 contra R\$ 931 das demais oito regiões somadas — uma diferença de **R\$ 66**, com **p = 0,09**. Pelo critério convencional de 5%, não é distinguível do acaso. E **168 dos 426 municípios não metropolitanos (39%) ganham mais que a mediana da RF1**.

**4. Desta vez, Porto Alegre não é o que sustenta o resultado**

Removendo a capital, a mediana da RF1 vai de R\$ 997 para R\$ 995 — praticamente nada. Com os 70 municípios da região, e não apenas os 20 polos da versão anterior, o peso de um único caso se dilui.

**5. A metropolitana é internamente heterogênea**

A RF1 reúne cinco COREDEs com medianas bem distintas:

| COREDE | Municípios | Mediana |
|---|---|---|
| Vale do Rio dos Sinos | 14 | R\$ 1.169 |
| Metropolitano Delta do Jacuí | 10 | R\$ 1.043 |
| Vale do Caí | 19 | R\$ 999 |
| Paranhana Encosta da Serra | 10 | R\$ 917 |
| Centro-Sul | 17 | R\$ 914 |

A distância entre o primeiro e o último é de R\$ 255 — maior que a diferença entre a RF1 e a média das demais regiões. "Região metropolitana" não descreve uma realidade econômica homogênea.

**6. O que o mapa mostra**

Os municípios do quintil superior não se concentram nos anéis internos. Formam uma mancha densa na Serra, a 100–200 km da capital, e aparecem dispersos no norte e na fronteira oeste. Dentro do anel de 100 km há municípios dos quintis mais baixos, sobretudo no Centro-Sul.

**Como esta conclusão difere da versão anterior**

A análise com 84 polos e agrupamentos incorretos colocava a RF1 em **6º de 9**, com p = 0,83 — sugerindo que a metropolitana era indistinguível do resto. Com a base completa e os agrupamentos oficiais, a metropolitana sobe para 2º e a diferença para as demais fica na fronteira da significância (p = 0,09).

**A resposta à pergunta não muda, mas o motivo muda.** Antes: "a metropolitana não se destaca". Agora: "a metropolitana se destaca um pouco, mas é superada pela Serra de forma clara". Nos dois casos existem regiões não metropolitanas com rendimento superior — só que agora sabemos qual, e com que margem.

**Limitações**

- **Corte único de 2010.** Nada aqui descreve tendência.
- **Falácia ecológica.** A unidade é o município; nada autoriza afirmar algo sobre indivíduos.
- **Sem ponderação populacional.** Cada município pesa 1: Porto Alegre (1,4 milhão) conta como Ipiranga do Sul (2,1 mil). Ponderar pela população deslocaria o resultado a favor das regiões com cidades grandes.
- **"Proximidade" medida de forma grosseira.** Os anéis do mapa usam distância em linha reta ao centroide de Porto Alegre, e o agrupamento por Região Funcional é administrativo, não de tempo de deslocamento.
- **A defasagem entre a malha (2022) e os dados (2010)** não afeta os cálculos, que não usam geometria — apenas a representação no mapa.

#### 6.4 Pergunta 4 — Existem municípios cuja renda dos residentes é incompatível com sua produção local, e o que os distingue dos demais de PIB igualmente baixo?

Esta é a única pergunta que usa o **terceiro dataset**, o PIB dos municípios do RS em 2010 (IBGE). Ele acrescenta três coisas que o Atlas não tem:

- **PIB per capita de 2010** — mesmo ano do Censo, então aqui não há defasagem temporal entre renda e produção, ao contrário do que acontece quando se usa a série 2013–2016.
- **População de cada município** — permite verificar se o porte separa os grupos.
- **Valor adicionado por setor** (agropecuária, indústria, serviços e administração pública) — permite testar a estrutura econômica, e não apenas indicadores sociais.

##### 6.4.1 Preparação: medindo a incompatibilidade

"Incompatível" precisa de uma referência — incompatível **com o quê**. A referência adotada é o próprio padrão estadual: a reta que liga PIB per capita a rendimento. O **resíduo** de cada município (renda observada menos renda prevista pelo seu PIB) é a medida de incompatibilidade.

| Resíduo | Leitura |
|---|---|
| **positivo** | ganha mais do que sua produção local faria esperar |
| **negativo** | produz muito, e a renda não fica com os residentes |

**Critério de corte:** `|resíduo| > 1 desvio-padrão`. É explicável, simétrico nas duas direções e não depende de escolher um percentil arbitrário.

**O recorte da segunda metade da pergunta.** "O que os distingue dos demais municípios de **PIB igualmente baixo**" exige restringir a comparação ao tercil inferior de PIB per capita — os dois grupos precisam produzir o mesmo por habitante para que a comparação isole a renda.

**Por que teste de permutação.** O grupo de interesse tem apenas 13 municípios. Com amostra desse tamanho, comparar medianas "no olho" não basta: qualquer diferença pode ser sorteio. O teste embaralha os rótulos dos grupos 20.000 vezes e mede quão extrema é a diferença observada — não supõe normalidade, o que importa com `n` pequeno e variáveis assimétricas.

```python
# ---------------------------------------------------------------
# Pergunta 4 — preparacao: renda incompativel com a producao local
# ---------------------------------------------------------------
# "Incompativel" precisa de uma referencia: incompativel COM O QUE se
# esperaria. A referencia e o proprio padrao estadual — a reta que liga PIB
# per capita a rendimento. O RESIDUO de cada municipio (renda observada
# menos renda prevista pelo seu PIB) e a medida de incompatibilidade.
#
#   residuo > 0  -> ganha mais do que sua producao local faria esperar
#   residuo < 0  -> produz muito e a renda nao fica com os residentes
#
# Corte: |residuo| > 1 desvio-padrao dos residuos. E explicavel, simetrico
# nas duas direcoes e nao depende de escolher um percentil arbitrario.
#
# A segunda metade da pergunta — "o que os distingue dos demais municipios
# de PIB igualmente baixo" — exige restringir a comparacao ao TERCIL
# INFERIOR de PIB. Comparar com o estado inteiro confundiria o efeito
# procurado com o simples fato de ter PIB maior.
#
# O dataset de PIB 2010 traz duas coisas que o Atlas nao tem e que mudam o
# que da para responder: a POPULACAO de cada municipio e a decomposicao do
# valor adicionado por SETOR. Alem disso, o PIB e do mesmo ano do Censo,
# entao nao ha defasagem temporal entre renda e producao.
import numpy as np

col_escolaridade = "% dos ocupados com ensino médio completo 2010"
col_renda = "Rendimento médio dos ocupados 2010"

SETORES = {
    "vab_agropecuaria_mil_reais": "Agropecuária",
    "vab_industria_mil_reais": "Indústria",
    "vab_servicos_exceto_adm_publica_mil_reais": "Serviços",
    "vab_administracao_publica_mil_reais": "Adm. pública",
}

# ---------------------------------------------------------------
# 1. Cruzamento Atlas x PIB 2010
# ---------------------------------------------------------------
pib_2010 = pd.read_csv(URL_PIB_2010)
print(f"Dataset de PIB 2010: {len(pib_2010)} municípios, {pib_2010.shape[1]} atributos")

base_atlas = df[~df["Territorialidades"].isin(["Brasil", "Rio Grande do Sul"])].copy()
for c in (col_renda, col_escolaridade):
    base_atlas[c] = pd.to_numeric(base_atlas[c], errors="coerce")

p4 = base_atlas.merge(pib_2010, on="Territorialidades", how="inner")
p4 = p4.dropna(subset=[col_renda, col_escolaridade, "pib_per_capita_2010"])
print(f"Cruzamento por 'Territorialidades': {len(p4)} municípios "
      f"(Atlas tem {len(base_atlas)}, PIB tem {len(pib_2010)})")

# participacao de cada setor no valor adicionado bruto
for coluna, nome in SETORES.items():
    p4[f"pct_{nome}"] = 100 * p4[coluna] / p4["vab_total_mil_reais"]

p4["pib_pc_mil"] = p4["pib_per_capita_2010"] / 1000

# ---------------------------------------------------------------
# 2. Renda esperada pelo PIB e residuo
# ---------------------------------------------------------------
# Ajuste em log do PIB: a variavel e muito assimetrica (de R$ 6,6 mil a
# R$ 220 mil por habitante), entao em nivel a reta ficaria refem dos poucos
# municipios-planta no topo.
with np.errstate(divide="ignore", over="ignore", invalid="ignore"):
    base_reg = np.column_stack([np.ones(len(p4)), np.log10(p4["pib_pc_mil"])])
    coefs_p4, *_ = np.linalg.lstsq(base_reg, p4[col_renda].to_numpy(float), rcond=None)
    p4["renda_prevista"] = base_reg @ coefs_p4

p4["residuo"] = p4[col_renda] - p4["renda_prevista"]
desvio_residuo = p4["residuo"].std()
r2_p4 = float(np.corrcoef(p4["renda_prevista"], p4[col_renda])[0, 1]) ** 2

print(f"\nRenda esperada = {coefs_p4[0]:.0f} + {coefs_p4[1]:.0f} × log10(PIB per capita em R$ mil)")
print(f"  R² = {r2_p4:.3f}  |  desvio-padrão do resíduo = R$ {desvio_residuo:,.0f}")

acima_1dp = p4["residuo"] > desvio_residuo
abaixo_1dp = p4["residuo"] < -desvio_residuo
print(f"\nMunicípios com renda incompatível com a produção local (|resíduo| > 1 desvio-padrão):")
print(f"  ganham MUITO MAIS do que produzem: {acima_1dp.sum()} de {len(p4)}")
print(f"  produzem muito e ganham POUCO:     {abaixo_1dp.sum()} de {len(p4)}")

print("\nMaiores resíduos NEGATIVOS (produção alta, renda baixa):")
print(p4.nsmallest(5, "residuo")[
    ["municipio", "pib_pc_mil", col_renda, "renda_prevista", "residuo", "pct_Indústria"]
].to_string(index=False, float_format=lambda v: f"{v:9.1f}"))

# ---------------------------------------------------------------
# 3. Recorte da pergunta: tercil inferior de PIB per capita
# ---------------------------------------------------------------
corte_pib = p4["pib_pc_mil"].quantile(1 / 3)
pib_baixo = p4[p4["pib_pc_mil"] <= corte_pib].copy()
pib_baixo["classe"] = np.where(
    pib_baixo["residuo"] > desvio_residuo,
    "Renda acima do esperado", "Renda compatível com o PIB",
)
print(f"\nTercil inferior de PIB per capita (≤ R$ {corte_pib * 1000:,.0f}/hab.): "
      f"{len(pib_baixo)} municípios")
print(pib_baixo["classe"].value_counts().to_string())

grupo_alto = pib_baixo[pib_baixo["classe"] == "Renda acima do esperado"]
grupo_resto = pib_baixo[pib_baixo["classe"] == "Renda compatível com o PIB"]

print("\nMunicípios de PIB baixo com renda acima do esperado:")
print(grupo_alto.nlargest(20, "residuo")[
    ["municipio", "pib_pc_mil", col_renda, "residuo", "populacao_censo_2010"]
].to_string(index=False, float_format=lambda v: f"{v:9.1f}"))

# ---------------------------------------------------------------
# 4. O que os distingue? Teste de permutacao para cada candidato
# ---------------------------------------------------------------
# Com n=13 de um lado, comparar medianas "no olho" nao basta: qualquer
# diferenca pode ser sorteio. O teste de permutacao embaralha os rotulos
# dos grupos e mede quao extrema e a diferenca observada. Nao supoe
# normalidade — importante com amostra pequena e variaveis assimetricas.
DISCRIMINANTES = [
    (col_escolaridade, "Ocupados com ensino médio completo (%)"),
    ("populacao_censo_2010", "População (habitantes, Censo 2010)"),
    ("pct_Serviços", "Serviços no valor adicionado (%)"),
    ("pct_Agropecuária", "Agropecuária no valor adicionado (%)"),
    ("pct_Indústria", "Indústria no valor adicionado (%)"),
    ("pct_Adm. pública", "Administração pública no valor adicionado (%)"),
]

N_PERMUTACOES = 20000
rng = np.random.default_rng(7)
resultados = []
for coluna, rotulo in DISCRIMINANTES:
    valores = pib_baixo[coluna].to_numpy(float)
    observada = np.median(grupo_alto[coluna]) - np.median(grupo_resto[coluna])
    difs = np.array([
        np.median(p[:len(grupo_alto)]) - np.median(p[len(grupo_alto):])
        for p in (rng.permutation(valores) for _ in range(N_PERMUTACOES))
    ])
    resultados.append({
        "indicador": rotulo,
        "grupo_alto": np.median(grupo_alto[coluna]),
        "grupo_resto": np.median(grupo_resto[coluna]),
        "diferenca": observada,
        "p_valor": float((np.abs(difs) >= abs(observada)).mean()),
    })

discriminantes = pd.DataFrame(resultados)
print(f"\nO que distingue os dois grupos (medianas e teste de permutação, "
      f"{N_PERMUTACOES:,} reamostras):".replace(",", "."))
# p-valor precisa de mais casas que as medianas, entao formata coluna a coluna
exibicao = discriminantes.copy()
for coluna in ("grupo_alto", "grupo_resto", "diferenca"):
    exibicao[coluna] = exibicao[coluna].map(lambda v: f"{v:,.1f}".replace(",", "."))
exibicao["p_valor"] = discriminantes["p_valor"].map(
    lambda v: "< 0,0001" if v < 0.0001 else f"{v:.4f}".replace(".", ",")
)
exibicao["distingue?"] = np.where(discriminantes["p_valor"] < 0.05, "sim", "NÃO")
print(exibicao.to_string(index=False))
print("\n  p < 0,05 indica diferença difícil de atribuir ao acaso.")
```

##### 6.4.4 Mapa — onde estão os municípios de renda incompatível

Os gráficos 4.1 e 4.2 identificam **quais** municípios fogem do padrão e **o que** os distingue. Este mostra **onde** eles estão — a dimensão que um eixo de dispersão não exibe.

Diferente do mapa da pergunta 3, aqui a cobertura é praticamente completa: o resíduo é calculado para os 496 municípios com dado de PIB em 2010. Só Pinto Bandeira fica de fora, por ter sido emancipado em 2013 (ver seção 4).

As classes **não** são quantis: são o próprio critério da análise, o mesmo desvio-padrão do resíduo usado em 6.4.1. Assim o mapa e a tabela contam a mesma história, sem introduzir um segundo corte arbitrário. O contorno preto marca os 13 municípios do tercil inferior de PIB com renda acima do esperado — o grupo que a segunda metade da pergunta investiga.

```python
# ---------------------------------------------------------------
# Mapa 4.3 - onde estao os municipios de renda incompativel
# ---------------------------------------------------------------
# Os graficos 4.1 e 4.2 identificam QUAIS municipios fogem do padrao e o
# que os distingue. Este mostra ONDE eles estao — a dimensao que um eixo
# de dispersao nao consegue exibir.
#
# Diferente do mapa da pergunta 3, aqui a cobertura e completa: o residuo
# e calculado para os 496 municipios com dado de PIB 2010. Apenas Pinto
# Bandeira fica de fora, por ter sido emancipado em 2013 e nao constar da
# base de 2010 (ver seção 4).
#
# As classes NAO sao quantis: sao o proprio criterio da analise, o mesmo
# desvio-padrao do residuo usado na secao 6.4.1. Assim o mapa e a tabela
# contam a mesma historia, sem um segundo corte arbitrario.
import matplotlib.pyplot as plt
from matplotlib.patches import Patch

CRS_MAPA = malha_rs.estimate_utm_crs()
base = malha_rs.to_crs(CRS_MAPA)

# junção por CÓDIGO do IBGE, não por nome: o dataset de PIB traz
# `codigo_ibge`, o que elimina o risco de erro de grafia
mapa_residuos = base.merge(
    p4.assign(CD_MUN=p4["codigo_ibge"].astype(str))[
        ["CD_MUN", "municipio", "residuo", "pib_pc_mil", col_renda]],
    on="CD_MUN", how="left", validate="one_to_one",
)
sem_dado = mapa_residuos["residuo"].isna()
print(f"Municípios mapeados: {(~sem_dado).sum()} | sem dado de PIB 2010: {sem_dado.sum()} "
      f"({', '.join(mapa_residuos.loc[sem_dado, 'NM_MUN'])})")

CLASSES = [
    (-float("inf"), -desvio_residuo, "#2166ac",
     f"Produz muito, renda baixa (< −1 dp)"),
    (-desvio_residuo, 0, "#a8c8e4", "Abaixo do esperado"),
    (0, desvio_residuo, "#f6b48a", "Acima do esperado"),
    (desvio_residuo, float("inf"), "#b2400a",
     f"Ganha muito mais do que produz (> +1 dp)"),
]
CINZA_SEM_DADO = "#d9d9d9"

fig, ax = plt.subplots(figsize=(10, 10.5))
mapa_residuos[sem_dado].plot(ax=ax, color=CINZA_SEM_DADO, edgecolor="white",
                             linewidth=0.25, zorder=1)
for minimo, maximo, cor, _ in CLASSES:
    fatia = mapa_residuos[(mapa_residuos["residuo"] > minimo) & (mapa_residuos["residuo"] <= maximo)]
    if not fatia.empty:
        fatia.plot(ax=ax, color=cor, edgecolor="white", linewidth=0.25, zorder=2)

# contorno nos 13 do tercil inferior de PIB com renda acima do esperado —
# o grupo que a segunda metade da pergunta investiga
codigos_destaque = set(grupo_alto["codigo_ibge"].astype(str))
realce = mapa_residuos[mapa_residuos["CD_MUN"].isin(codigos_destaque)]
realce.boundary.plot(ax=ax, color="#0b0b0b", linewidth=1.3, zorder=4)

for nome_mun, deslocamento in {"Pelotas": (-70, -30), "Viamão": (66, -10),
                               "Uruguaiana": (-30, 34), "Triunfo": (74, 16)}.items():
    linha = mapa_residuos[mapa_residuos["municipio"] == nome_mun]
    if linha.empty:
        continue
    ponto = linha.geometry.iloc[0].centroid
    ax.annotate(f"{nome_mun}\nR\\$ {linha['residuo'].iloc[0]:+,.0f}".replace(",", "."),
                xy=(ponto.x, ponto.y), xytext=deslocamento, textcoords="offset points",
                fontsize=8.5, color="#1a1a1a", zorder=6, ha="center",
                bbox=dict(boxstyle="round,pad=0.2", fc="white", ec="#c9c9c4", alpha=0.9),
                arrowprops=dict(arrowstyle="-", color="#52514e", linewidth=0.7))

ax.set_axis_off()
ax.set_title(
    "Onde estão os municípios cuja renda não acompanha a produção local?\n"
    f"Resíduo do rendimento em relação ao esperado pelo PIB per capita — "
    f"{int((~sem_dado).sum())} municípios do RS",
    fontsize=13, pad=14,
)

ax.legend(
    handles=[Patch(facecolor=c, edgecolor="white", label=r) for _, _, c, r in CLASSES]
    + [Patch(facecolor=CINZA_SEM_DADO, edgecolor="white", label="Sem dado de PIB em 2010"),
       Patch(facecolor="none", edgecolor="#0b0b0b", linewidth=1.3,
             label=f"PIB baixo e renda acima do esperado (n={len(realce)})")],
    frameon=False, loc="lower left", fontsize=8.5,
    title=f"Resíduo (1 desvio-padrão = R\\$ {desvio_residuo:,.0f})".replace(",", "."),
    title_fontsize=9,
)

fig.text(
    0.01, 0.02,
    "Fontes: Atlas do Desenvolvimento Humano no Brasil (Pnud, Ipea, FJP), Censo 2010 (IBGE); "
    "PIB dos municípios 2010 (IBGE); malha municipal 2022 (IBGE).\n"
    "Resíduo = rendimento observado − rendimento previsto pelo log do PIB per capita. Junção feita pelo código do "
    "IBGE, não pelo nome.\n"
    f"Projeção {CRS_MAPA.name}.",
    fontsize=8, color="#8a8a86",
)

plt.tight_layout()
# plt.savefig("mapa4.3_residuos_renda_producao.png", dpi=150, bbox_inches="tight")
plt.show()
```

##### 6.4.5 Interpretação

**Resposta curta: sim, existem — 147 dos 496 municípios (30%). E o que distingue os de renda alta entre os de PIB baixo é serem centros urbanos de serviços, não municípios agrícolas pequenos.**

**1. A incompatibilidade aparece nas duas direções**

| Direção | n | Exemplos extremos |
|---|---|---|
| Produzem muito, renda baixa | **75** | Pinhal da Serra (−R\$ 854), Pinhal Grande (−R\$ 707), Triunfo (−R\$ 650) |
| Ganham muito mais do que produzem | **72** | Mariano Moro (+R\$ 439), Herval (+R\$ 430), Pelotas (+R\$ 429) |

Os extremos negativos têm em comum a indústria: Pinhal da Serra tem **86,6%** do valor adicionado na indústria, Entre Rios do Sul 88,3%, Triunfo 70,2%. São municípios-planta — produzem muito por habitante, e o valor não fica com quem mora ali.

**2. Entre os de PIB baixo, o que separa os dois grupos**

No tercil inferior de PIB per capita (≤ R\$ 13.284/hab., 166 municípios), 13 têm renda acima do esperado. Comparados aos outros 153 — que produzem o mesmo por habitante:

| Indicador | Renda acima (n=13) | Renda compatível (n=153) | Diferença | p |
|---|---|---|---|---|
| Rendimento | R\$ 1.194 | R\$ 729 | +R\$ 465 | — |
| Ocupados com ensino médio | 38,0% | 23,0% | +15,0 p.p. | < 0,0001 |
| **População** | **41.585** | **5.285** | **≈ 8×** | < 0,0001 |
| **Serviços no VAB** | **45,6%** | **26,4%** | +19,2 p.p. | < 0,0001 |
| **Agropecuária no VAB** | **9,5%** | **36,9%** | −27,4 p.p. | < 0,0001 |
| Indústria no VAB | 12,3% | 5,7% | +6,6 p.p. | 0,0040 |
| Administração pública no VAB | 25,9% | 28,0% | −2,1 p.p. | **0,2382** |

**3. O retrato que emerge**

Os 13 são **centros urbanos de serviços**: Pelotas (328 mil habitantes), Viamão (239 mil), Alvorada (196 mil), Uruguaiana (125 mil), Bagé (117 mil), Santiago, Canela, Tramandaí. Cidades grandes, com economia de serviços, cujo PIB por habitante é baixo justamente porque o denominador populacional é grande — mas onde as pessoas ganham bem acima do que essa razão sugere.

Os outros 153 são o oposto: municípios pequenos (mediana de 5.285 habitantes) com mais de um terço do valor adicionado na agropecuária.

Ou seja, "PIB per capita baixo" agrupa duas realidades opostas. Num caso é sinal de **porte urbano** — muita gente dividindo uma produção que não é medida no município; no outro é sinal de **economia agrícola de pequeno porte**. O indicador não distingue as duas, e a renda dos moradores sim.

**4. Um resultado negativo que vale registrar**

A administração pública era a explicação mais intuitiva — cidades que vivem de emprego público. **Não se sustenta:** é o único candidato testado que não distingue os grupos (25,9% contra 28,0%, p = 0,24). A diferença está em serviços privados e porte urbano, não no setor público.

**5. Conexão com as outras perguntas**

A pergunta 1 mostrou que escolaridade explica metade da variação de renda; a 3, que a proximidade da capital não explica. Esta acrescenta que o PIB per capita também não: o modelo tem **R² = 0,21**, ou seja, quase 80% da variação de renda entre municípios não tem relação com o quanto cada um produz por habitante.

**Limitações**

- **O resíduo depende do modelo.** Com R² = 0,21, a maior parte do resíduo é simplesmente o que o PIB não explica, e não um "desvio" no sentido forte. Trocar a forma funcional muda quem entra na lista.
- **Selecionar pelos extremos garante regressão à média.** Numa reamostragem, parte dos 13 sairia da lista. O padrão de grupo é mais confiável que a composição exata.
- **`n` = 13.** Os testes de permutação sustentam as diferenças de grupo, mas as medianas do grupo pequeno são sensíveis a poucos casos.
- **Escolaridade e renda são correlacionadas** (r = 0,70, pergunta 1), então um grupo definido por renda alta tende a aparecer com escolaridade alta em parte por construção. Os discriminantes estruturais — população e composição setorial — não têm esse problema.
- **Falácia ecológica:** a unidade é o município. Nada aqui autoriza afirmar algo sobre indivíduos.
- **Sem dados de deslocamento pendular**, a explicação de que parte desses moradores trabalha fora do município fica plausível, mas não verificada.
