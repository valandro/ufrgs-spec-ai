```python
# Preparação do ambiente
import pandas as pd

# Carregamento dos dados
arquivo = "../dataset_principal.csv" 
df = pd.read_csv(arquivo)

# Exploração inicial dos dados
print("INFORMAÇÕES DA BASE DE DADOS")
print("=" * 70)

print(f"Número de registros: {df.shape[0]}")
print(f"Número de atributos: {df.shape[1]}")

print("\nRESUMO DOS REGISTROS")
print("=" * 70)

print(df.head())
```

**Base de Dados:** Não foi aplicada nenhuma amostragem, a análise considera a totalidade dos 496 municípios do Rio Grande do Sul presentes na base, além dos registros agregados de Brasil e Rio Grande do Sul (usados apenas como referência, não incluídos na análise por unidade municipal).

```python
# As duas primeiras linhas correspondem ao Brasil e ao Rio Grande do Sul.
# Elas serão excluídas da análise das unidades individuais.
df_municipios = df.iloc[2:].reset_index(drop=True)

# Verificar colunas totalmente vazias
colunas_vazias = df_municipios.columns[df_municipios.isna().all()].tolist()

# Resumo
print("\nATRIBUTOS COMPLETAMENTE VAZIOS (MUNICÍPIOS)")
print("=" * 70)

print(f"Total de registros para análise (excluindo registros do Brasil e RS): {df_municipios.shape[0]}")
print(f"Total de atributos: {df_municipios.shape[1]}")
print(f"Atributos totalmente vazios (municípios): {len(colunas_vazias)}")

# filtrar as colunas totalmente vazias do DataFrame
df_municipios = df_municipios.drop(columns=colunas_vazias)
print(f"Total de atributos com algum dado (municípios): {df_municipios.shape[1]}")
```

```python
#Filtrar os atributos com percentual de nulos maior que 90%
df_filtrado = df_municipios.loc[:, (df_municipios.isna().sum() / len(df_municipios)) * 100 <= 95]

resumo = []

print("CARACTERÍSTICAS DOS ATRIBUTOS COM PERCENTUAL DE NULOS <= 95%")
print("=" * 70)

for coluna in df_filtrado.columns:

    tipo = df_filtrado[coluna].dtype
    nulos = df_filtrado[coluna].isna().sum()
    percentual_nulos = (nulos / len(df_filtrado)) * 100

    resumo.append({
        "Atributo": coluna,
        "Tipo": str(tipo),
        "Nulos": nulos,
        "% nulos": round(percentual_nulos, 2)
    })

resumo_df = pd.DataFrame(resumo)

#ordenar o DataFrame pelo percentual de nulos em ordem decrescente
resumo_df = resumo_df.sort_values(by="% nulos", ascending=False)
print("\n")
print(resumo_df.to_string(index=False))

print(f"\nNúmero de registros filtrados: {df_filtrado.shape[0]}")
print(f"Número de atributos filtrados: {df_filtrado.shape[1]}")
```

```python
colunas_interesse = [
    "Territorialidades",
    "% dos ocupados com ensino médio completo 2010",
    "Rendimento médio dos ocupados 2010",
    "% dos ocupados sem rendimento 2010",
    "% dos ocupados com rendimento de até 1 salário mínimo 2010",
    "% dos ocupados com rendimento de até 2 salários mínimo 2010",
    "% dos ocupados com rendimento de até 3 salários mínimo 2010",
    "% dos ocupados com rendimento de até 5 salários mínimo 2010"
]

df_filtrado = df_filtrado[colunas_interesse]

# Transformar a coluna "% dos ocupados com ensino médio completo 2010" em numérica
df_filtrado["% dos ocupados com ensino médio completo 2010"] = pd.to_numeric(
    df_filtrado["% dos ocupados com ensino médio completo 2010"], errors="coerce"
)

colunas_com_nulos = [
    "% dos ocupados com rendimento de até 1 salário mínimo 2010",
    "% dos ocupados sem rendimento 2010",
    "Rendimento médio dos ocupados 2010",
    "% dos ocupados com ensino médio completo 2010",
]

print("\nREGISTROS COM NULOS NOS ATRIBUTOS DE INTERESSE")
print("=" * 70)

linhas_nulas = df_municipios[df_municipios[colunas_com_nulos].isnull().any(axis=1)][["Territorialidades"] + colunas_com_nulos]
print(linhas_nulas.to_string(index=False))

```

```python
# Filtrar os registros com nulos nos atributos de interesse
df_filtrado = df_filtrado.dropna(subset=colunas_com_nulos).reset_index(drop=True)

print("ESTATÍSTICAS DOS ATRIBUTOS DE INTERESSE")
print("=" * 70)
print(f"\nMunicípios para análise: {len(df_filtrado)}")

print(df_filtrado.describe().T.to_string())
```

```python
import unicodedata

# Checagem de duplicidades em df_filtrado

# 1. Linhas 100% idênticas (todas as colunas iguais)
linhas_duplicadas = df_filtrado[df_filtrado.duplicated(keep=False)]
if len(linhas_duplicadas) > 0:
    print(linhas_duplicadas[["Territorialidades"]].to_string())

# 2. Nome de município repetido (comparação exata)
contagem_nomes = df_filtrado["Territorialidades"].value_counts()
nomes_repetidos = contagem_nomes[contagem_nomes > 1]
if len(nomes_repetidos) > 0:
    print(nomes_repetidos)

# 3. Nome "quase" igual — normaliza (tira acento, espaço extra, maiúscula/minúscula)
#    e compara de novo. Pega erro de digitação que a comparação exata não pega.
def normalizar(texto):
    texto = str(texto).strip().lower()
    texto = unicodedata.normalize("NFKD", texto).encode("ascii", "ignore").decode("ascii")
    return " ".join(texto.split())

df_filtrado = df_filtrado.copy()
df_filtrado["_nome_normalizado"] = df_filtrado["Territorialidades"].apply(normalizar)
contagem_normalizado = df_filtrado["_nome_normalizado"].value_counts()
possiveis_variacoes = contagem_normalizado[contagem_normalizado > 1]

if len(possiveis_variacoes) > 0:
    for nome_norm in possiveis_variacoes.index:
        variantes = df_filtrado[df_filtrado["_nome_normalizado"] == nome_norm]["Territorialidades"].tolist()
        print(f"  {nome_norm!r} aparece como: {variantes}")

df_filtrado = df_filtrado.drop(columns=["_nome_normalizado"])

# 4. Resumo
print("ANÁLISE DE DUPLICIDADES")
print("=" * 60)
print(f"Linhas totalmente idênticas:              {len(linhas_duplicadas)}")
print(f"Nomes de município exatamente repetidos:   {len(nomes_repetidos)}")
print(f"Possíveis variações do mesmo nome:         {len(possiveis_variacoes)}")
if len(linhas_duplicadas) == 0 and len(nomes_repetidos) == 0 and len(possiveis_variacoes) == 0:
    print("\nNenhuma duplicidade encontrada nos 3 níveis checados.")
```

```python
import matplotlib.pyplot as plt

col_escolaridade = "% dos ocupados com ensino médio completo 2010"
col_renda = "Rendimento médio dos ocupados 2010"

atributos = [
    (col_escolaridade, "% de ocupados com ensino médio completo (2010)", "%"),
    (col_renda, "Rendimento médio dos ocupados (2010)", "R$"),
]
 
fig, axes = plt.subplots(1, 2, figsize=(11, 4.5))
 
for ax, (coluna, titulo, unidade) in zip(axes, atributos):
    serie = df_filtrado[coluna].dropna()
 
    contagens, _, _ = ax.hist(serie, bins=20, edgecolor="white", linewidth=0.6, zorder=3)
 
    media = serie.mean()
    ax.axvline(media, color="#0b0b0b", linestyle="--", linewidth=1, alpha=0.6, zorder=4)
 
    # headroom acima da barra mais alta, pra caber o rótulo da média sem sobrepor
    y_max = contagens.max()
    ax.set_ylim(0, y_max * 1.18)
 
    rotulo = f"R$ {media:,.2f}" if unidade == "R$" else f"{media:.1f}%"
    ax.text(
        media, y_max * 1.05,
        f" média = {rotulo}",
        fontsize=8.5, color="#3a3a38", va="bottom", ha="left",
    )
 
    ax.set_title(titulo, fontsize=11, pad=10)
    ax.set_xlabel(unidade if unidade == "%" else "R$")
    ax.set_ylabel("Nº de municípios")
 
    ax.spines[["top", "right"]].set_visible(False)
    ax.set_axisbelow(True)
    ax.grid(axis="y", linestyle="-", linewidth=0.5, zorder=0)
 
fig.suptitle(
    "Distribuição dos municípios do RS - escolaridade e renda dos ocupados",
    fontsize=13, y=1.03,
)
 
plt.tight_layout()
plt.show()
```

Os dois histogramas apresentam distribuição unimodal e assimetria à direita, indicando maior concentração dos municípios em valores baixos ou intermediários e uma quantidade menor de municípios com valores mais elevados.

Na escolaridade, a maior concentração está entre 20% e 40% de ocupados com ensino médio completo. A média (29,7%) e a mediana (29,1%) são próximas, indicando uma assimetria relativamente moderada.

Na renda, os valores estão mais concentrados entre R$ 750 e R$ 1.250. A média (R$ 956,61) é superior à mediana (R$ 936,76), evidenciando uma influência maior dos municípios com rendimentos elevados, que chegam a aproximadamente R$ 2.250.

De forma geral, os histogramas indicam maior dispersão da renda entre os municípios do que da escolaridade, com a presença de alguns municípios que se afastam da concentração observada.
