Apresentem:

```python
# Preparação do ambiente
import pandas as pd

# Carregamento dos dados
arquivo = "../dataset_principal.csv" 
df = pd.read_csv(arquivo)

# Exploração inicial dos dados
print("INFORMAÇÕES DA BASE DE DADOS")
print("=" * 70)

print(f"Número de registros: {df.shape[0]}")
print(f"Número de atributos: {df.shape[1]}")

print("\nRESUMO DOS REGISTROS")
print("=" * 70)

print(df.head())
```

```python
# As duas primeiras linhas correspondem ao Brasil e ao Rio Grande do Sul.
# Elas serão excluídas da análise das unidades individuais.
df_municipios = df.iloc[2:].reset_index(drop=True)

# Verificar colunas totalmente vazias
colunas_vazias = df_municipios.columns[df_municipios.isna().all()].tolist()

print("COLUNAS TOTALMENTE VAZIAS")
print("=" * 70)

if len(colunas_vazias) == 0:
    print("Não existem colunas totalmente vazias.")
else:
    for coluna in colunas_vazias:
        print(f"- {coluna}")
```

```python
# Resumo
print("\nRESUMO")
print("=" * 70)

print(f"Total de registros para análise (excluindo registros do Brasil e RS): {df_municipios.shape[0]}")
print(f"Total de atributos: {df_municipios.shape[1]}")
print(f"Atributos totalmente vazios (municípios): {len(colunas_vazias)}")

# filtrar as colunas totalmente vazias do DataFrame
df_municipios = df_municipios.drop(columns=colunas_vazias)
print(f"Total de atributos com algum dado (municípios): {df_municipios.shape[1]}")
```

```python
# Caracterização dos atributos

resumo = []

print("CARACTERÍSTICAS DOS ATRIBUTOS")
print("=" * 70)

for coluna in df_municipios.columns:

    tipo = df_municipios[coluna].dtype
    nulos = df_municipios[coluna].isna().sum()
    percentual_nulos = (nulos / len(df_municipios)) * 100

    resumo.append({
        "Atributo": coluna,
        "Nulos": nulos,
        "% nulos": round(percentual_nulos, 2)
    })

resumo_df = pd.DataFrame(resumo)

#ordenar o DataFrame pelo percentual de nulos em ordem decrescente
resumo_df = resumo_df.sort_values(by="% nulos", ascending=False)
print("\n")
print(resumo_df.to_string(index=False))
```

```python
#Filtrar os atributos com percentual de nulos maior que 90%
df_filtrado = df_municipios.loc[:, (df_municipios.isna().sum() / len(df_municipios)) * 100 <= 95]

resumo = []

print("CARACTERÍSTICAS DOS ATRIBUTOS COM PERCENTUAL DE NULOS <= 95%")
print("=" * 70)

for coluna in df_filtrado.columns:

    tipo = df_filtrado[coluna].dtype
    nulos = df_filtrado[coluna].isna().sum()
    percentual_nulos = (nulos / len(df_filtrado)) * 100

    resumo.append({
        "Atributo": coluna,
        "Tipo": str(tipo),
        "Nulos": nulos,
        "% nulos": round(percentual_nulos, 2)
    })

resumo_df = pd.DataFrame(resumo)

#ordenar o DataFrame pelo percentual de nulos em ordem decrescente
resumo_df = resumo_df.sort_values(by="% nulos", ascending=False)
print("\n")
print(resumo_df.to_string(index=False))

print(f"\nNúmero de registros filtrados: {df_filtrado.shape[0]}")
print(f"Número de atributos filtrados: {df_filtrado.shape[1]}")
```

```python
#converter '% dos ocupados com ensino médio completo 2010' para numérico
atributo = "% dos ocupados com ensino médio completo 2010"

df_filtrado[atributo] = pd.to_numeric(
    df_filtrado[atributo],
    errors="coerce"
)

numericos = df_filtrado.select_dtypes(include="number").columns

print("ESTATÍSTICAS DOS ATRIBUTOS NUMÉRICOS")
print("=" * 70)

print(df_filtrado[numericos].describe().T.to_string())
```

```python
import math
import matplotlib.pyplot as plt
import textwrap


# ============================================================
# 6. DISTRIBUIÇÃO DOS ATRIBUTOS NUMÉRICOS NÃO BINÁRIOS
# ============================================================

# Atributos numéricos binários
binarios = [
    coluna for coluna in numericos
    if coluna.startswith("Existência de coleta seletiva")
]

print("\nATRIBUTOS BINÁRIOS")
print("-" * 70)

for coluna in binarios:
    print(coluna)

# Atributos numéricos que serão usados nos histogramas
numericos_histograma = [
    coluna for coluna in numericos
    if coluna not in binarios
]

print(
    f"\nAtributos numéricos para histogramas: "
    f"{len(numericos_histograma)}"
)

print("\nHISTOGRAMAS DOS ATRIBUTOS NUMÉRICOS NÃO BINÁRIOS")
print("=" * 70)
numero_colunas = 3
numero_linhas = math.ceil(len(numericos_histograma) / numero_colunas)

fig, axes = plt.subplots(
    numero_linhas,
    numero_colunas,
    figsize=(15, 4 * numero_linhas)
)

# Transformar axes em lista para facilitar o tratamento
axes = axes.flatten()

for i, coluna in enumerate(numericos_histograma):

    df_filtrado[coluna].dropna().plot(
        kind="hist",
        bins=20,
        ax=axes[i],
        edgecolor="black"
    )

    titulo = "\n".join(textwrap.wrap(coluna, width=35))

    axes[i].set_title(titulo, fontsize=10)
    axes[i].set_xlabel("Valor")
    axes[i].set_ylabel("Frequência")


# Remover espaços vazios
for i in range(len(numericos_histograma), len(axes)):
    fig.delaxes(axes[i])

plt.tight_layout()
plt.show()

```

```python
#Grafico de dispersão e regressão linear entre "% dos ocupados com ensino médio completo 2010" e "Rendimento médio dos ocupados 2010"
import numpy as np

# Atributos
x = "% dos ocupados com ensino médio completo 2010"
y = "Rendimento médio dos ocupados 2010"

# Remover valores ausentes
dados = df_filtrado[[x, y]].dropna()

# Dados para a regressão
X = dados[x]
Y = dados[y]

# Coeficientes da reta: Y = aX + b
a, b = np.polyfit(X, Y, 1)

# Valores da reta
reta = a * X + b

# Gráfico
plt.figure(figsize=(10, 6))

plt.scatter(
    X,
    Y,
    alpha=0.6,
    label="Municípios"
)

# Ordenar X para desenhar a reta corretamente
ordem = np.argsort(X)

plt.plot(
    X.iloc[ordem],
    reta.iloc[ordem],
    "r--",
    linewidth=2,
    label="Regressão linear"
)

plt.xlabel("% dos ocupados com ensino médio completo 2010")
plt.ylabel("Rendimento médio dos ocupados 2010 (R$)")
plt.title("Relação entre escolaridade e rendimento dos ocupados - 2010")

plt.grid(True, alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()

# Correlação de Pearson
correlacao = X.corr(Y)

print("\nRESULTADOS")
print("=" * 70)

print(f"Número de municípios: {len(dados)}")
print(f"Coeficiente de correlação de Pearson: {correlacao:.3f}")
```

É possível identificar no Gráfico X **(Atualizar)** uma correlação positiva entre a porcentagem de ocupados com ensino médio completo e o rendimento médio dos ocupados. Em geral, os municípios com maior porcentagem de ocupados com ensino médio completo tendem a apresentar maiores rendimentos médios.

O gráfico também permite identificar alguns possíveis valores discrepantes (outliers). Observa-se apenas um município com mais de 60% dos ocupados com ensino médio completo. O rendimento médio desse município está bem acima do valor estimado pela linha de regressão, indicando um desvio positivo em relação à tendência observada nos demais municípios.

Além disso, alguns municípios apresentam renda média elevada mesmo com menor escolaridade, indicando que outros fatores também podem estar relacionados à renda. Isso sugere a necessidade de investigar outras características econômicas e sociais dos municípios.
