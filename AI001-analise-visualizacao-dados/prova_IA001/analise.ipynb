### Preparação — idioma das avaliações como indicador de origem

As duas perguntas dependem de saber se o hóspede é estrangeiro, e nenhum arquivo traz o país de origem. Uso o **idioma do comentário** em `reviews.csv` como indicador aproximado: português = doméstico; qualquer outro idioma identificado = estrangeiro.

O idioma é identificado por contagem de palavras funcionais de cada língua (por exemplo "não", "muito" e "com" para o português; "the", "and" e "was" para o inglês). Vence o idioma com mais ocorrências; comentários com empate ou com menos de duas ocorrências ficam como "indefinido" e saem da análise.

Como `reviews.csv` tem 404 MB, o script `preparar_dados.py` o resume em dois arquivos pequenos, que acompanham a entrega:

- `reviews_idioma_anuncio.csv`: uma linha por anúncio, com as avaliações domésticas e estrangeiras dos últimos 12 meses (jul/2025 a jun/2026).
- `reviews_idioma_bairro_mes.csv`: avaliações domésticas e estrangeiras por mês, bairro e tipo de acomodação, desde jul/2022.

Se os dois arquivos já existem na pasta, o notebook apenas os lê. Se faltarem, a célula abaixo chama `preparar_dados.py`, que baixa `reviews.csv.gz` de https://insideairbnb.com/get-the-data/ (coleta de 24/06/2026, cerca de 155 MB) e os recalcula. O mesmo pode ser feito pelo terminal com `python preparar_dados.py`.

```python
# Dados derivados de reviews.csv: idioma do comentário como indicador da origem do hóspede
import os

ARQ_ANUNCIO = 'reviews_idioma_anuncio.csv'        # 1 linha por anúncio (jul/2025 a jun/2026)
ARQ_BAIRRO_MES = 'reviews_idioma_bairro_mes.csv'  # 1 linha por mês x bairro x tipo de acomodação (desde jul/2022)

# Os agregados acompanham a entrega. Se faltarem, preparar_dados.py baixa reviews.csv.gz
# do Inside Airbnb (cerca de 155 MB), classifica o idioma dos comentários e os recalcula.
if not (os.path.exists(ARQ_ANUNCIO) and os.path.exists(ARQ_BAIRRO_MES)):
    import preparar_dados
    preparar_dados.garantir_agregados()

idioma_anuncio = pd.read_csv(ARQ_ANUNCIO)
idioma_bairro_mes = pd.read_csv(ARQ_BAIRRO_MES)

# Base por anúncio: listings + avaliações classificadas dos últimos 12 meses
anuncios = df.merge(idioma_anuncio, left_on='id', right_on='listing_id', how='left')
anuncios[['n_domestico', 'n_estrangeiro']] = anuncios[['n_domestico', 'n_estrangeiro']].fillna(0).astype(int)
anuncios['n_classificadas'] = anuncios['n_domestico'] + anuncios['n_estrangeiro']
anuncios['pct_estrangeiro'] = (anuncios['n_estrangeiro'] / anuncios['n_classificadas']).where(anuncios['n_classificadas'] > 0)

# Paleta e rótulos usados em todos os gráficos (a mesma cor sempre representa a mesma série)
COR_UNICA = '#2a78d6'
ORIGENS = ['Doméstico (português)', 'Estrangeiro (outros idiomas)']
CORES_ORIGEM = ['#2a78d6', '#eb6834']

total = anuncios[['n_domestico', 'n_estrangeiro']].sum()
print(f'Anúncios com avaliação classificada nos últimos 12 meses: {(anuncios["n_classificadas"] > 0).sum():,}')
print(f'Avaliações classificadas: {total.sum():,} | estrangeiras: {total["n_estrangeiro"] / total.sum():.1%}')
```

### Gráfico 1 — Preço da diária de imóveis inteiros por bairro *(Pergunta 1)*

**Justificativa do tipo:** gráfico de pontos com intervalo (mediana e 1º–3º quartil), ordenado pela mediana. Barras exigiriam eixo linear partindo de zero, e o Joá (mediana de R$ 7.704) comprimiria os outros 31 bairros em uma faixa ilegível. Um boxplot com valores atípicos seria dominado pelos extremos (o preço máximo é R$ 574.013). O intervalo interquartil mostra também a dispersão dentro do bairro, que a mediana sozinha esconde.

**Decisões de tratamento:**
- **Filtragem:** apenas imóveis inteiros (81% dos anúncios), para não misturar quartos e imóveis completos; excluídos os 3.318 imóveis inteiros sem preço; mantidos os 32 bairros (de 154) com pelo menos 100 anúncios, que reúnem 94,9% dos imóveis inteiros com preço.
- **Escala:** logarítmica no eixo de preço, porque as medianas vão de R$ 228 a R$ 7.704.
- **Moeda:** o arquivo não informa a moeda de `price`; assumo reais, porque o Inside Airbnb publica os preços na moeda local do anúncio.
- **Extremos:** nenhum valor foi excluído; mediana e quartis são pouco sensíveis a valores extremos, o que a média não é (média de R$ 878 contra mediana de R$ 453 na base toda).

```python
# Gráfico 1 — preço da diária por bairro (mediana e intervalo interquartil)
MIN_ANUNCIOS_BAIRRO = 100

inteiros = anuncios[(anuncios['room_type'] == 'Entire home/apt') & anuncios['price'].notna()]
preco_bairro = (
    inteiros.groupby('neighbourhood')['price']
    .agg(anuncios='size', q1=lambda s: s.quantile(0.25), mediana='median', q3=lambda s: s.quantile(0.75))
    .reset_index()
)
preco_bairro = preco_bairro[preco_bairro['anuncios'] >= MIN_ANUNCIOS_BAIRRO]

eixo_preco = alt.Scale(type='log', domain=[100, 20000])
base1 = alt.Chart(preco_bairro).encode(
    y=alt.Y('neighbourhood:N', sort=alt.EncodingSortField('mediana', order='descending'), title='Bairro'),
    tooltip=[
        alt.Tooltip('neighbourhood:N', title='Bairro'),
        alt.Tooltip('anuncios:Q', title='Anúncios', format=','),
        alt.Tooltip('q1:Q', title='1º quartil (R$)', format=',.0f'),
        alt.Tooltip('mediana:Q', title='Mediana (R$)', format=',.0f'),
        alt.Tooltip('q3:Q', title='3º quartil (R$)', format=',.0f'),
    ],
)
faixa1 = base1.mark_rule(strokeWidth=3, opacity=0.35, color=COR_UNICA).encode(
    x=alt.X('q1:Q', scale=eixo_preco, title='Preço da diária (R$, escala logarítmica)',
            axis=alt.Axis(values=[100, 200, 500, 1000, 2000, 5000, 10000, 20000], format=',.0f')),
    x2='q3:Q',
)
ponto1 = base1.mark_point(size=70, filled=True, opacity=1, color=COR_UNICA).encode(x=alt.X('mediana:Q', scale=eixo_preco))

grafico1 = (faixa1 + ponto1).properties(
    width=560, height=alt.Step(17),
    title=alt.TitleParams(
        'Preço da diária de imóveis inteiros por bairro',
        subtitle=[f'Ponto = mediana; barra = do 1º ao 3º quartil. Bairros com pelo menos {MIN_ANUNCIOS_BAIRRO} anúncios com preço.'],
        anchor='start',
    ),
)
grafico1
```

### Gráfico 2 — Avaliações em idioma estrangeiro por faixa de preço dentro do bairro *(Pergunta 1)*

**Justificativa do tipo:** gráfico de linhas com pontos sobre um eixo ordinal (quartis de preço). A faixa de preço tem ordem, e a inclinação de cada linha mostra diretamente se o percentual de estrangeiros cresce com o preço. Uma linha por bairro permite comparar a tendência entre bairros com níveis muito diferentes (Ipanema e Centro), o que barras agrupadas mostrariam com menos clareza. Um gráfico de dispersão com os 11.503 anúncios teria sobreposição excessiva.

**Decisões de tratamento:**
- **Quartis dentro do bairro:** a faixa de preço é calculada separadamente em cada bairro, para que a comparação seja entre anúncios da mesma localização e o resultado não repita apenas a diferença entre bairros do Gráfico 1.
- **Filtragem:** imóveis inteiros com preço e com pelo menos 5 avaliações de idioma identificado entre jul/2025 e jun/2026, para que o percentual de cada anúncio não dependa de 1 ou 2 comentários. Restam 11.503 dos 20.774 imóveis inteiros com preço desses bairros (55,4%).
- **Bairros:** os quatro com mais anúncios da cidade (Copacabana, Ipanema, Barra da Tijuca e Centro); limitar a quatro séries mantém as cores distinguíveis, e o resultado vale para esses bairros.
- **Idioma:** comentários sem idioma identificado (8,3%, em geral muito curtos) ficam fora do cálculo.
- **Escala:** eixo vertical linear a partir de zero.

```python
# Gráfico 2 — avaliações estrangeiras por faixa de preço dentro do mesmo bairro
MIN_AVALIACOES_ANUNCIO = 5
BAIRROS_G2 = ['Ipanema', 'Copacabana', 'Barra da Tijuca', 'Centro']
CORES_G2 = ['#2a78d6', '#eb6834', '#1baf7a', '#eda100']
FAIXAS = ['Q1 (mais baratos)', 'Q2', 'Q3', 'Q4 (mais caros)']

com_avaliacoes = anuncios[
    (anuncios['room_type'] == 'Entire home/apt') & anuncios['price'].notna()
    & (anuncios['n_classificadas'] >= MIN_AVALIACOES_ANUNCIO) & anuncios['neighbourhood'].isin(BAIRROS_G2)
].copy()
# O quartil é calculado dentro de cada bairro, para comparar anúncios da mesma localização
com_avaliacoes['faixa_preco'] = com_avaliacoes.groupby('neighbourhood')['price'].transform(
    lambda preco: pd.qcut(preco, 4, labels=FAIXAS)
).astype(str)

estrangeiro_faixa = (
    com_avaliacoes.groupby(['neighbourhood', 'faixa_preco'])
    .agg(pct_estrangeiro=('pct_estrangeiro', 'mean'), anuncios=('id', 'size'), preco_mediano=('price', 'median'))
    .reset_index()
)

base2 = alt.Chart(estrangeiro_faixa).encode(
    x=alt.X('faixa_preco:N', sort=FAIXAS, title='Faixa de preço dentro do próprio bairro (quartis)',
            axis=alt.Axis(labelAngle=0)),
    y=alt.Y('pct_estrangeiro:Q', title='Avaliações em idioma estrangeiro (média por anúncio)',
            scale=alt.Scale(domain=[0, 0.8]), axis=alt.Axis(format='.0%')),
    color=alt.Color('neighbourhood:N', title='Bairro', sort=BAIRROS_G2,
                    scale=alt.Scale(domain=BAIRROS_G2, range=CORES_G2)),
    tooltip=[
        alt.Tooltip('neighbourhood:N', title='Bairro'),
        alt.Tooltip('faixa_preco:N', title='Faixa de preço'),
        alt.Tooltip('pct_estrangeiro:Q', title='Avaliações estrangeiras', format='.1%'),
        alt.Tooltip('preco_mediano:Q', title='Preço mediano (R$)', format=',.0f'),
        alt.Tooltip('anuncios:Q', title='Anúncios', format=','),
    ],
)
linhas2 = base2.mark_line(strokeWidth=2, point=alt.OverlayMarkDef(size=70, filled=True, opacity=1))
rotulos2 = (
    base2.transform_filter(alt.datum.faixa_preco == FAIXAS[-1])
    .mark_text(align='left', dx=10, fontSize=12, color='#52514e')
    .encode(text='neighbourhood:N', color=alt.value('#52514e'))
)

grafico2 = (linhas2 + rotulos2).properties(
    width=520, height=340,
    title=alt.TitleParams(
        'Nos quatro maiores bairros, anúncios mais caros recebem mais avaliações em idioma estrangeiro',
        subtitle=[f'Quatro bairros com mais anúncios; imóveis inteiros com pelo menos {MIN_AVALIACOES_ANUNCIO} '
                  'avaliações classificadas entre jul/2025 e jun/2026.'],
        anchor='start',
    ),
)
grafico2
```

### Gráfico 3 — Concentração das avaliações por bairro, por origem do hóspede *(Pergunta 2)*

**Justificativa do tipo:** curva de concentração (participação acumulada com os bairros ordenados do mais ao menos avaliado). A pergunta é sobre "em quantos bairros" cada público se distribui, e a curva responde isso diretamente: basta ler em que posição cada linha cruza a referência de 80%. Barras por bairro mostrariam onde cada grupo fica, mas não quantos bairros são necessários.

**Decisões de tratamento:**
- **Ranking próprio de cada grupo:** os bairros são ordenados separadamente para cada origem; a posição 2, por exemplo, é Centro para o público doméstico e Ipanema para o estrangeiro (o nome aparece ao passar o cursor).
- **Filtragem:** todos os tipos de acomodação; avaliações com idioma identificado entre jul/2025 e jun/2026 (416.887 de 454.645); eixo limitado aos 15 primeiros bairros, onde está a diferença entre as curvas.
- **Escala:** linear de 0 a 100%, com linha de referência em 80%.

```python
# Gráfico 3 — curva de concentração das avaliações por bairro, por origem do hóspede
MAX_BAIRROS = 15

por_bairro = anuncios.groupby('neighbourhood')[['n_domestico', 'n_estrangeiro']].sum()
curvas = []
for coluna, origem in zip(['n_domestico', 'n_estrangeiro'], ORIGENS):
    # cada grupo tem o seu próprio ranking de bairros, do mais ao menos avaliado
    ordenado = por_bairro[coluna].sort_values(ascending=False)
    participacao = ordenado / ordenado.sum()
    curvas.append(pd.DataFrame({
        'origem': origem,
        'posicao': range(1, len(ordenado) + 1),
        'bairro': ordenado.index,
        'participacao': participacao.values,
        'acumulado': participacao.cumsum().values,
    }))
concentracao = pd.concat(curvas)
concentracao = concentracao[concentracao['posicao'] <= MAX_BAIRROS]

base3 = alt.Chart(concentracao).encode(
    x=alt.X('posicao:Q', title='Número de bairros (do mais ao menos avaliado por cada grupo)',
            scale=alt.Scale(domain=[1, MAX_BAIRROS], nice=False),
            axis=alt.Axis(values=list(range(1, MAX_BAIRROS + 1)), format='d', grid=False)),
    y=alt.Y('acumulado:Q', title='Participação acumulada nas avaliações do grupo',
            scale=alt.Scale(domain=[0, 1]), axis=alt.Axis(format='.0%')),
    color=alt.Color('origem:N', title='Origem do hóspede', sort=ORIGENS,
                    scale=alt.Scale(domain=ORIGENS, range=CORES_ORIGEM)),
    tooltip=[
        alt.Tooltip('origem:N', title='Origem'),
        alt.Tooltip('posicao:Q', title='Posição no ranking'),
        alt.Tooltip('bairro:N', title='Bairro nesta posição'),
        alt.Tooltip('participacao:Q', title='Participação do bairro', format='.1%'),
        alt.Tooltip('acumulado:Q', title='Acumulado', format='.1%'),
    ],
)
linhas3 = base3.mark_line(strokeWidth=2, point=alt.OverlayMarkDef(size=60, filled=True, opacity=1))
referencia3 = alt.Chart(pd.DataFrame({'y': [0.8], 'texto': ['80% das avaliações']})).mark_rule(
    strokeDash=[4, 4], color='#8a8984'
).encode(y='y:Q')
texto3 = alt.Chart(pd.DataFrame({'x': [MAX_BAIRROS], 'y': [0.8], 'texto': ['80% das avaliações']})).mark_text(
    align='right', dy=-7, fontSize=11, color='#52514e'
).encode(x='x:Q', y='y:Q', text='texto:N')

grafico3 = (referencia3 + texto3 + linhas3).properties(
    width=560, height=340,
    title=alt.TitleParams(
        'Avaliações estrangeiras se concentram em menos bairros que as domésticas',
        subtitle=['Todos os tipos de acomodação; avaliações classificadas entre jul/2025 e jun/2026.'],
        anchor='start',
    ),
)
grafico3
```

### Gráfico 4 — Participação da orla da Zona Sul nas avaliações ao longo do tempo *(Pergunta 2)*

**Justificativa do tipo:** série temporal em linhas, uma por origem. A segunda parte da pergunta é sobre mudança no tempo, e linhas sobre um eixo temporal contínuo mostram tendência e sazonalidade; as duas séries ficam no mesmo eixo porque têm a mesma unidade.

**Decisões de tratamento:**
- **Indicador:** percentual das avaliações de cada grupo em anúncios de Copacabana, Ipanema, Leblon e Leme. Usar percentual, e não contagem, separa a concentração do crescimento do volume total de avaliações.
- **Agregação:** trimestral, para reduzir a oscilação mensal sem esconder a sazonalidade.
- **Período:** de jul/2022 a jun/2026, que exclui os anos de pandemia (em 2021 as avaliações em espanhol caíram para 2,4% do total).
- **Escala:** linear de 0 a 100%, para não exagerar visualmente as variações.
- **Cuidado na leitura:** a base só contém anúncios ativos em junho de 2026. Anúncios que saíram do mercado não aparecem nos períodos antigos, então a série mistura mudança de comportamento dos hóspedes com mudança na composição da oferta.

```python
# Gráfico 4 — evolução da participação da orla da Zona Sul nas avaliações, por origem
ORLA_ZONA_SUL = ['Copacabana', 'Ipanema', 'Leblon', 'Leme']

serie = idioma_bairro_mes.copy()
serie['trimestre'] = pd.PeriodIndex(serie['mes'], freq='Q').start_time
serie['na_orla'] = serie['neighbourhood'].isin(ORLA_ZONA_SUL)
serie = serie.melt(
    id_vars=['trimestre', 'na_orla'], value_vars=['n_domestico', 'n_estrangeiro'],
    var_name='origem', value_name='avaliacoes',
)
serie['origem'] = serie['origem'].map(dict(zip(['n_domestico', 'n_estrangeiro'], ORIGENS)))

orla = (
    serie.groupby(['trimestre', 'origem'])
    .apply(lambda g: pd.Series({
        'pct_orla': g.loc[g['na_orla'], 'avaliacoes'].sum() / g['avaliacoes'].sum(),
        'avaliacoes': g['avaliacoes'].sum(),
    }), include_groups=False)
    .reset_index()
)

base4 = alt.Chart(orla).encode(
    x=alt.X('trimestre:T', title='Trimestre da avaliação (mês de início)',
            axis=alt.Axis(format='%m/%Y', tickCount=8, grid=False)),
    y=alt.Y('pct_orla:Q', title='Avaliações em anúncios da orla da Zona Sul',
            scale=alt.Scale(domain=[0, 1]), axis=alt.Axis(format='.0%')),
    color=alt.Color('origem:N', title='Origem do hóspede', sort=ORIGENS,
                    scale=alt.Scale(domain=ORIGENS, range=CORES_ORIGEM)),
    tooltip=[
        alt.Tooltip('origem:N', title='Origem'),
        alt.Tooltip('trimestre:T', title='Início do trimestre', format='%m/%Y'),
        alt.Tooltip('pct_orla:Q', title='Na orla da Zona Sul', format='.1%'),
        alt.Tooltip('avaliacoes:Q', title='Avaliações no trimestre', format=','),
    ],
)
linhas4 = base4.mark_line(strokeWidth=2, point=alt.OverlayMarkDef(size=50, filled=True, opacity=1))
rotulos4 = (
    base4.transform_filter(alt.datum.trimestre == alt.expr.toDate(orla['trimestre'].max().isoformat()))
    .mark_text(align='left', dx=8, fontSize=12)
    .encode(text=alt.Text('pct_orla:Q', format='.0%'), color=alt.value('#52514e'))
)

grafico4 = (linhas4 + rotulos4).properties(
    width=560, height=340,
    title=alt.TitleParams(
        'Participação da orla da Zona Sul nas avaliações, por origem do hóspede',
        subtitle=['Orla da Zona Sul = Copacabana, Ipanema, Leblon e Leme. Avaliações classificadas de jul/2022 a jun/2026.'],
        anchor='start',
    ),
)
grafico4
```

### Conclusões — Pergunta 1

**Resposta:** *O preço da diária varia muito entre os bairros. Entre os 32 bairros com pelo menos 100 imóveis inteiros anunciados, a diária mediana vai de R$ 228 (Campo Grande) a R$ 7.704 (Joá). E sim: entre imóveis inteiros com no mínimo 5 avaliações, nos 4 maiores bairros, quanto mais caro o anúncio, maior a proporção de avaliações em idioma estrangeiro. Do quartil mais barato ao mais caro dentro de cada bairro, essa proporção passa de 39,5% para 57,2% em Copacabana, de 56,3% para 72,8% em Ipanema, de 16,0% para 52,6% na Barra da Tijuca e de 13,1% para 24,7% no Centro.*
